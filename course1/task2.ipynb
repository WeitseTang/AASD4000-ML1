# Task 2 - Web Scraper Notebook

Scrape a Medium article and save the extracted text to a file.

Article URL: https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7

## 1. Import libraries

```python
import os
import sys
import requests
import re
from bs4 import BeautifulSoup
```

## 2. Function: get the HTML source of the Medium article

```python
def get_page():
    global url

    url = input("Enter url of a medium article: ")

    # handling possible error
    if not re.match(r'https?://medium.com/', url):
        print('Please enter a valid website, or make sure it is a medium article')
        sys.exit(1)

    res = requests.get(url)
    res.raise_for_status()
    soup = BeautifulSoup(res.text, 'html.parser')
    return soup
```

## 3. Function: clean HTML tags from text

```python
def clean(text):
    rep = {"<br>": "\n", "<br/>": "\n", "<li>": "\n"}
    rep = dict((re.escape(k), v) for k, v in rep.items())
    pattern = re.compile("|".join(rep.keys()))
    text = pattern.sub(lambda m: rep[re.escape(m.group(0))], text)
    text = re.sub('\<(.*?)\>', '', text)
    return text
```

## 4. Function: collect text from all paragraphs

```python
def collect_text(soup):
    text = f'url: {url}\n\n'
    para_text = soup.find_all('p')
    print(f"paragraphs text = \n {para_text}")
    for para in para_text:
        text += f"{para.text}\n\n"
    return text
```

## 5. Function: save text to a .txt file

```python
def save_file(text):
    if not os.path.exists('./scraped_articles'):
        os.mkdir('./scraped_articles')
    name = url.split("/")[-1]
    print(name)
    fname = f'scraped_articles/{name}.txt'

    with open(fname, 'w', encoding='utf-8') as f:
        f.write(text)

    print(f'File saved in directory {fname}')
```

## 6. Run the scraper

When prompted, paste:
`https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7`

```python
text = collect_text(get_page())
save_file(text)
```
